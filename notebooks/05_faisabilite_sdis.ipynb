## Pourquoi ce notebook ?
On veut vérifier le contenu réel des données avant de retenir les clés, les périodes et les indicateurs. Les noms de colonnes ci-dessous sont des hypothèses, pas un schéma certifié.

## 1. Chargement et traçabilité

Avant de lancer l'analyse, renseigner la ressource exacte, sa date de téléchargement, sa licence et son empreinte SHA-256 dans le registre des extractions. Ne jamais commiter de données SDIS sensibles.

```python
from pathlib import Path
import hashlib
import pandas as pd

SOURCE = Path("data/restricted/interventions.csv")  # À adapter après récupération de la ressource
if not SOURCE.exists():
    print(f"Fichier absent : {SOURCE}. Aucune analyse de données réalisée.")
    df = None
else:
    print("SHA-256 :", hashlib.sha256(SOURCE.read_bytes()).hexdigest())
    df = pd.read_csv(SOURCE, sep=None, engine="python", dtype=str)
    print("Dimensions :", df.shape)
    display(df.head())
```

## 2. Profilage initial

Champs envisagés, à confirmer : `date_alerte, date_arrivee, identifiant_intervention`.

**Question adressée à Steve (retour attendu) :** les identifiants et le grain observés permettent-ils de répondre aux questions Q5, Q6 ?

```python
if df is not None:
    profil = pd.DataFrame({
        "type": df.dtypes.astype(str),
        "manquants": df.isna().sum(),
        "pct_manquants": (100 * df.isna().mean()).round(2),
        "distincts": df.nunique(dropna=True),
    })
    display(profil.sort_values("pct_manquants", ascending=False))
    print("Doublons exacts :", int(df.duplicated().sum()))
else:
    print("Profilage en attente du fichier source.")
```

## 3. Visualisation des manquants

Ce graphique est produit uniquement sur des observations chargées. Il ne représente pas un résultat métier tant que les colonnes et la couverture ne sont pas confirmées.

```python
import matplotlib.pyplot as plt
if df is not None and len(df.columns):
    taux = (df.isna().mean() * 100).sort_values(ascending=False).head(15)
    ax = taux.plot(kind="bar", figsize=(10, 4), title="Valeurs manquantes par colonne (%)")
    ax.set_ylabel("%")
    plt.tight_layout()
    plt.show()
else:
    print("Graphique en attente des données.")
```
